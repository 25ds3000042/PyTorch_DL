```python
import torch
import torch.nn as nn
```

```python
# 1. Create training data
X = torch.tensor([
    [[1.0], [2.0], [3.0], [4.0]],
    [[2.0], [3.0], [4.0], [5.0]],
    [[3.0], [4.0], [5.0], [6.0]],
    [[4.0], [5.0], [6.0], [7.0]]
])
```

```python
y = torch.tensor([
    [5.0],
    [6.0],
    [7.0],
    [8.0]
])
```

```python
# 2. Build RNN model
class SimpleRNN(nn.Module):

    def __init__(self):

        super().__init__()

        self.rnn = nn.RNN(
            input_size=1,
            hidden_size=12,
            num_layers=1,
            batch_first=True
        )

        self.fc = nn.Linear(12, 1)

    def forward(self, x):

        output, hidden = self.rnn(x)

        output = output[:, -1, :]

        prediction = self.fc(output)

        return prediction
```

```python
# 3. Initialize model
model = SimpleRNN()
```

```python
# 4. Loss function
criterion = nn.MSELoss()
```

```python
# 5. Optimizer
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.01
)
```

```python
# 6. Training loop
for epoch in range(500):

    prediction = model(X)

    loss = criterion(prediction, y)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    if (epoch + 1) % 50 == 0:
        print(
            f"Epoch {epoch+1}, Loss: {loss.item():.4f}"
        )

```

```python
# 7. Prediction
test = torch.tensor([[[5.0], [6.0], [7.0], [8.0]]])

model.eval()

with torch.no_grad():
    result = model(test)

print("Predicted next number:", result.item())
```
