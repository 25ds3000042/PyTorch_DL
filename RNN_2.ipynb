```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
sentence = ["I", "ate", "an", "apple"]
```

```python
vocab = sorted(set(sentence))

print(vocab)
```

```python
word_to_idx = {word:i for i, word in enumerate(vocab)}
idx_to_word = {i:word for word,i in word_to_idx.items()}

print(word_to_idx)
```

```python
encoded = [word_to_idx[word] for word in sentence]

print(encoded)
```

```python
inputs = [
    [word_to_idx["I"]],
    [word_to_idx["I"], word_to_idx["ate"]],
    [word_to_idx["I"], word_to_idx["ate"], word_to_idx["an"]]
]

targets = [
    word_to_idx["ate"],
    word_to_idx["an"],
    word_to_idx["apple"]
]

print(inputs)
print(targets)
```
